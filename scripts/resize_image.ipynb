```python
import os
import numpy as np
import ants
import matplotlib.pyplot as plt

from glom_pop import dataio

base_dir = dataio.get_config_file()['base_dir']

# meanbrain_tag = '20241021'
meanbrain_tag = '20241021_tall'
brain_dir = os.path.join(base_dir, 'mean_brain', meanbrain_tag)

```

```python
# Load the NIfTI image
image = ants.image_read(os.path.join(brain_dir, 'TSeries-20240925-001_anatomical.nii'))

# Get the shape of the image
shape = image.shape  # This returns (x, y, z)

# Define the cropping range
# Here, we want to keep the full range of x and y, but drop the last layer of z
crop_region = (slice(0, shape[0]), slice(0, shape[1]), slice(0, shape[2] - 1))
```

```python
# Apply the cropping
# cropped_image = image[crop_region]
# Define the cropping region: full x, full y, and z range from 0 to 65 (excluding 66)
cropped_image = ants.crop_indices(image, lowerind=[0, 0, 0], upperind=[image.shape[0]-1, image.shape[1]-1, image.shape[2]-2])
```

```python

# Save the cropped image
ants.image_write(cropped_image, os.path.join(brain_dir, 'TSeries-20240925-001_anatomical_65z.nii'))
```
